```python
import pandas as pd

# Load the dataset
df = pd.read_csv("enhanced_zomato_dataset_clean.csv")

# Count the number of restaurants/items for each cuisine
cuisine_count = df["Cuisine"].value_counts()

# Top 5 most common cuisines
top_5 = cuisine_count.head(5)

print("Top 5 Most Common Cuisines:")
print(top_5)


# Checking the average rating of top 5 cuisines
top_cuisines = top_5.index

top_data = df[df["Cuisine"].isin(top_cuisines)]

average_rating = top_data.groupby("Cuisine")["Average_Rating"].mean()

print("\nAverage Rating of Top 5 Cuisines:")
print(average_rating.sort_values(ascending=False))
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv("enhanced_zomato_dataset_clean.csv")


# Rating Distribution
plt.figure(figsize=(8, 5))

sns.histplot(df["Average_Rating"], bins=10, kde=True)

plt.title("Distribution of Restaurant Ratings")
plt.xlabel("Average Rating")
plt.ylabel("Number of Restaurants")
plt.show()


# Top 10 Cuisines
top_cuisines = df["Cuisine"].value_counts().head(10)

plt.figure(figsize=(10, 5))

sns.barplot(x=top_cuisines.values, y=top_cuisines.index)

plt.title("Top 10 Most Common Cuisines")
plt.xlabel("Number of Restaurants")
plt.ylabel("Cuisine")
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv("enhanced_zomato_dataset_clean.csv")


# Restaurant with maximum votes
max_votes = df["Votes"].max()

print("Restaurant with Maximum Votes:")
print(df[df["Votes"] == max_votes][["Restaurant_Name", "Votes", "Average_Rating"]])


# Restaurant with minimum votes
min_votes = df["Votes"].min()

print("\nRestaurant with Minimum Votes:")
print(df[df["Votes"] == min_votes][["Restaurant_Name", "Votes", "Average_Rating"]])


# Correlation between votes and rating
correlation = df["Votes"].corr(df["Average_Rating"])

print("\nCorrelation between Votes and Rating:")
print(correlation)


# Heatmap
corr_data = df[["Votes", "Average_Rating"]].corr()

plt.figure(figsize=(6, 4))

sns.heatmap(corr_data, annot=True, cmap="coolwarm")

plt.title("Correlation Between Votes and Average Rating")
plt.show()
```
